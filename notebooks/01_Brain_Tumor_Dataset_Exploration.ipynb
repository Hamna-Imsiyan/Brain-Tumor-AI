```python

import kagglehub

dataset_path = kagglehub.dataset_download(
    "briscdataset/brisc2025"
)

print("Dataset downloaded to:")
print(dataset_path)
```

```python

from pathlib import Path
from itertools import islice

folder = Path(dataset_path)

print("First 30 items found:")

for item in islice(folder.rglob("*"), 30):
    print(item.relative_to(folder))
```
